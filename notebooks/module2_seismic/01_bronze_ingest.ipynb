## Sanity check: raw waveform plot

Visualize a couple of raw traces before trusting anything downstream —
per the dev doc's "visualize before modeling" step.

```python
import matplotlib.pyplot as plt
import pandas as pd
import local_sample

meta = pd.read_csv("../../data/module2_vibration/bronze/stead_sample_metadata.csv")
trace_name = meta.iloc[0]["trace_name"]
trace = local_sample.get_waveform("stead", trace_name)  # (channels, samples)

plt.figure(figsize=(10, 3))
plt.plot(trace[0])
plt.title(f"STEAD sample trace: {trace_name} (raw, unfiltered)")
plt.xlabel("sample")
plt.ylabel("amplitude")
plt.show()
```
